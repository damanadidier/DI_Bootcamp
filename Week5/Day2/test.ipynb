```python
import numpy as np

# 1. Setup: Simulated Data
# X = square footage (in hundreds), y = price (in thousands)
X = np.array([1, 2, 3, 4, 5])
y = np.array([150, 200, 250, 300, 350])

# 2. Initialization
weight = 0.0      # Starting guess
learning_rate = 0.01
iterations = 1000
n = len(X)

# 3. Gradient Descent Loop
for i in range(iterations):
    # Predict (y = weight * X)
    prediction = weight * X
    
    # Calculate Error (difference between predicted and actual)
    error = prediction - y
    
    # Calculate Gradient: The derivative of the error function
    # It tells us the slope of the error at our current 'weight'
    gradient = (2/n) * np.sum(error * X)
    
    # Update Weight: Move against the gradient
    weight = weight - (learning_rate * gradient)

print(f"Optimal weight found: {weight:.2f}")
print(f"Prediction for size 6: {weight * 6:.2f}")
```
