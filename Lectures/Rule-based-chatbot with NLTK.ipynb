# Création d'un Chatbot Basé sur des Règles (Rule-based Chatbot) avec NLTK

Ce notebook illustre la méthode traditionnelle de création de chatbot. Contrairement aux modèles d'IA modernes (LLM), un chatbot à base de règles s'appuie sur le filtrage de mots-clés et la correspondance de motifs (pattern matching).

```python
import nltk
from nltk.chat.util import Chat, reflections

# Téléchargement des ressources nécessaires
nltk.download('punkt')
```

## Définition des Règles (Patterns and Responses)

Nous utilisons des expressions régulières pour faire correspondre l'entrée de l'utilisateur à une réponse prédéfinie.

```python
pairs = [
    [
        r"mon nom est (.*)",
        ["Bonjour %1, comment puis-je vous aider aujourd'hui ?",]
    ],
    [
        r"quel est ton nom ?",
        ["Je suis un assistant virtuel basé sur des règles.",]
    ],
    [
        r"comment vas-tu ?",
        ["Je vais bien, merci ! Et vous ?",]
    ],
    [
        r"(hi|hello|bonjour|salut)",
        ["Bonjour !", "Salut !", "Heureux de vous voir."]
    ],
    [
        r"(.*) (ville|météo) (.*)",
        ["Je ne peux pas encore vérifier la météo en direct, mais j'espère qu'il fait beau !",]
    ],
    [
        r"quitter",
        ["Au revoir ! Passez une excellente journée.",]
    ],
]

# Affichage des réflexions (permet de transformer 'je' en 'vous', etc.)
print("Exemple de réflexions NLTK :", reflections)
```

## Lancement du Chatbot

Nous créons une instance de `Chat` avec nos paires de règles.

```python
def start_chat():
    print("Chatbot: Bonjour ! Tapez 'quitter' pour sortir.")
    chat = Chat(pairs, reflections)
    chat.converse()

# Note: Dans un environnement interactif comme Colab,
# converse() attendra une saisie dans un champ texte.
if __name__ == "__main__":
    start_chat()
```
