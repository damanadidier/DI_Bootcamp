```text
# Maria owns a small coffee shop in the city.
# Maria asks herself each morning:
# "How much coffee should I prepare today?"

- If she makes too much, it goes to waste.
- If she makes too little, customers leave disappointed.

--------------------------------------------------

Mon–Fri

Temperature outside
Cups Sold

        Temperature (°C) | Cups Sold

Mon            15        |    120
Tue            18        |    105
Wed            10        |    145
Thu            12        |    135
Fri            20        |     95

--------------------------------------------------

> If tomorrow is 8°C,
> how many cups should Maria prepare?
```

```text
Linear regression draws the best possible straight line
through your scattered data points.
```

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
````

````python
temperature = np.array([15, 18, 10, 12, 20, 8, 16, 14, 11, 19, 13, 17, 9, 22])
cups_sold = np.array([120, 105, 145, 135, 95, 155, 115, 125, 140, 100, 130, 110, 148, 88])
````

````python
plt.scatter(temperature, cups_sold)
plt.xlabel('Temperature')
plt.ylabel('Cups Sold')
````

````python
slope, intercept, r_value, p_value, std_err = stats.linregress(temperature, cups_sold) # linear regression (X, y)

line = slope * temperature + intercept
````

````python
plt.scatter(temperature, cups_sold)
plt.plot(temperature, line, color='red')
plt.xlabel('Temperature')
plt.ylabel('Cups Sold')
````

````python
slope
# SLOPE tells us that for every 1C increase the temperature – Maria sells `-4.86` FEWER cups of coffee
````

````python
intercept
# If it were 0C (freezing) – Maria would rhetorically sell 193 cups (This is the starting point of our line)
````

```text
y = mx + b    # EVERY LINE has a simple equation

x = input (temperature)

m = slope

b = intercept
```

```text
Temperature matters.

Each degree change in temperature affects sales
by approximately 4–5 cups.

This gives Maria the power to predict demand
and prepare accordingly.
```

````python
def predict_cups(temperature: int | float):
    prediction = slope * temperature + intercept
    return prediction
````

````python
tomorrow_temperature = 8
predict_cups(tomorrow_temperature)
````
