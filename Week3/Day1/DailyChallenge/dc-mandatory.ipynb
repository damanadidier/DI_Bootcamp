# Daily Challenge: Global Weather Temperature Analysis

**Scenario:** You are a data analyst working with a dataset of global weather. Your task is to analyze temperature trends and visualize the results.

**Tasks:**
1. **Data Preparation** — Generate synthetic monthly temperatures (12 months × 10 cities) ranging from -5 to 35 °C and store them in a Pandas DataFrame.
2. **Data Analysis** — Compute each city's annual average temperature and find the warmest and coldest cities.
3. **Data Visualization** — Plot the monthly temperature trends for every city.
4. **Report** — Summarize the findings.

## 1. Data Preparation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Reproducible results
np.random.seed(42)

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
cities = [f'City {i}' for i in range(1, 11)]

# Generate synthetic temperatures: 10 cities (rows) x 12 months (columns)
data = np.random.uniform(low=-5, high=35, size=(len(cities), len(months)))

# Build the DataFrame: cities as index, months as columns
df = pd.DataFrame(data, index=cities, columns=months).round(1)
df
```

## 2. Data Analysis

```python
# Annual average temperature for each city (mean across the months -> axis=1)
annual_avg = df.mean(axis=1).round(2)
annual_avg.name = 'Annual Avg (°C)'
annual_avg
```

```python
# City with the highest and lowest annual average temperature
hottest_city = annual_avg.idxmax()
coldest_city = annual_avg.idxmin()

print(f'Hottest city: {hottest_city} ({annual_avg.max():.2f} °C)')
print(f'Coldest city: {coldest_city} ({annual_avg.min():.2f} °C)')
```

## 3. Data Visualization

```python
plt.figure(figsize=(14, 7))

# One line per city across the 12 months
for city in df.index:
    plt.plot(df.columns, df.loc[city], marker='o', label=city)

plt.title('Monthly Temperature Trends for Each City', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='center left', bbox_to_anchor=(1.0, 0.5))
plt.tight_layout()
plt.show()
```

## 4. Summary Report


- The dataset contains **synthetic monthly temperature data** for **10 cities** across **12 months**.
- Annual average temperatures were calculated for each city, revealing that:
  - The hottest city is **City 3** with an average temperature of **30.25 °C**.
  - The coldest city is **City 7** with an average temperature of **10.75 °C**.
