```python
# Exercise 1 : Matrix Operations
# Instructions
# In this exercise, you’ll work with a 3x3 matrix. Here’s a brief explanation of the concepts:

# Determinant: The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus.
# Inverse of a Matrix: The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.
# Create a 3x3 matrix and perform the following operations:

# Calculate the determinant.
# Find the inverse of the matrix.
import numpy as np
# Create a 3x3 matrix
matrix = np.array([[1, 2, 3],
                   [0, 1, 4],
                   [5, 6, 0]])
# Calculate the determinant
determinant = np.linalg.det(matrix)
print("Determinant:", determinant)
# Find the inverse of the matrix
if determinant != 0:
    inverse_matrix = np.linalg.inv(matrix)
    print("Inverse of the matrix:\n", inverse_matrix)
else:
    print("The matrix is singular and does not have an inverse.")

    
```

```python
# Exercise 6 : Basic Hypothesis Testing
# Instructions
# Create a sample dataset to test the effectiveness of a new training program on employee productivity:


# import numpy as np

# # Productivity scores of employees before the training program
# productivity_before = np.random.normal(loc=50, scale=10, size=30)

# # Productivity scores of the same employees after the training program
# productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# # Your task is to formulate a hypothesis regarding the training program's effectiveness 
# # and test it using basic statistical functions in NumPy.


# Given a dataset, formulate a simple hypothesis and test it using basic statistical functions in NumPy.
import numpy as np
from scipy import stats
# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)
# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)
# Formulate a hypothesis:
# Null Hypothesis (H0): The training program has no effect on employee productivity (mean difference = 0).
# Alternative Hypothesis (H1): The training program has a positive effect on employee productivity (mean difference > 0).
# Calculate the mean difference
mean_difference = np.mean(productivity_after - productivity_before)
print("Mean Difference:", mean_difference)
# Perform a one-sample t-test to test the hypothesis
t_statistic, p_value = stats.ttest_rel(productivity_after, productivity_before)
print("T-statistic:", t_statistic)
print("P-value:", p_value)
# Conclusion
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis. The training program is effective.")
else:
    print("Fail to reject the null hypothesis. The training program is not effective.")



```

```python
# Exercise 8 : Time Series Data Manipulation
# Instructions
# Generate time series data for the year 2023. Demonstrate slicing for the following intervals:

# January to March
# April to June
# July to September
# October to December
# Generate a time series data for a specific period and demonstrate how to slice this data for different intervals.
import pandas as pd
# Generate time series data for the year 2023
date_range = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')
time_series_data = pd.Series(data=np.random.rand(len(date_range)), index=date_range)
# Slice for January to March
jan_mar = time_series_data['2023-01-01':'2023-03-31']
print("January to March:\n", jan_mar)
# Slice for April to June
apr_jun = time_series_data['2023-04-01':'2023-06-30']
print("April to June:\n", apr_jun)
# Slice for July to September
jul_sep = time_series_data['2023-07-01':'2023-09-30']
print("July to September:\n", jul_sep)
# Slice for October to December
oct_dec = time_series_data['2023-10-01':'2023-12-31']
print("October to December:\n", oct_dec)

```

```python
# Exercise 10 : Basic Visualization
# Instructions
# Use Matplotlib to visualize a simple dataset created with NumPy (e.g., a line graph of random numbers).

import numpy as np
import matplotlib.pyplot as plt
# Create a simple dataset with NumPy
x = np.arange(0, 10, 0.1)
y = np.sin(x) + np.random.normal(scale=0.5, size=len(x))
# Visualize the dataset using Matplotlib
plt.figure(figsize=(10, 5))
plt.plot(x, y, label='Random Sine Wave', color='blue')
plt.title('Line Graph of Random Sine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid()
plt.show()

```
