## SciPy vs NumPy

### SciPy

Used for:

- Advanced statistical tests (t-test, ANOVA, chi-square)
- Probability distributions (normal, binomial, Poisson, etc.)
- Statistical modeling and regression analysis
- Optimization
- Numerical integration
- Signal processing

### NumPy

Provides:

- Arrays and vectorized operations
- Mathematical computations
- Basic statistics:
  - Mean
  - Median
  - Standard deviation
  - Percentiles

---

## Quick Comparison

| NumPy | SciPy |
|--------|--------|
| Focuses on data and numerical computation | Focuses on scientific and statistical analysis |
| Arrays and basic statistics | Advanced statistical methods |
| Descriptive analytics | Inferential analytics |

### Simple Rule of Thumb

```text
NumPy = "What IS the data?" (descriptive)

SciPy = "What can we CONCLUDE from the data?"
         (inferential statistics)
```

````python
import numpy as np
import seaborn as sns
import scipy
import matplotlib.pyplot as plt
````

````python
tips_df = sns.load_dataset("tips")
tips_df.head()
````

````python
tip_amount = tips_df['tip'].values
tip_amount[:5]
````

````python
describe = scipy.stats.describe(tip_amount)
````

````python
describe.nobs # number of observations
````

````python
describe.minmax # minimum and maximum values
````

````python
describe.mean # mean value
````

````python
describe.variance # variance of the data
````

````python
describe.skewness # skewness of the data shows the asymmetry of the distribution
````

````python
describe.kurtosis # kurtosis of the data messures the "tailedness" of the data distribution
````

````python
# plt.hist(tip_amount, bins=20)
# plt.axvline(describe.mean, color='red', linestyle='dashed', linewidth=1)
sns.histplot(tip_amount, bins=20, kde=True)
````

Standard deviation is a measure of how spread out the numbers in a dataset are. It is calculated as the square root of the variance, which is the average of the squared differences from the mean. A low standard deviation indicates that the data points are close to the mean, while a high standard deviation indicates that the data points are more spread out.

````python
mean = describe.mean
````

````python
std = np.std(tip_amount)
````

````python
plt.hist(tip_amount, bins=20)
plt.axvline(describe.mean, color='red', linestyle='dashed', linewidth=1)
plt.axvline(describe.mean + std, color='orange', linestyle='dashed', linewidth=1)
````

````python
print(f"{(mean - std):.2f} < tip_amount < {(mean + std):.2f}")
print(describe.mean)
````
