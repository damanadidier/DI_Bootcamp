# Tracer votre premier graphique

Nous devons d’abord importer la bibliothèque `matplotlib`.

```python
import matplotlib
```

Traçons maintenant notre premier graphique ! :)

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 4, 9, 5, 3])
plt.show()
```

Oui, c'est aussi simple que d'appeler la fonction `plot` avec certaines données, puis d'appeler la fonction `show` !

**Remarque** :

> Matplotlib peut générer des graphiques à l'aide de diverses bibliothèques graphiques backend, telles que Tk, wxPython, etc. Lorsque vous exécutez Python à l'aide de la ligne de commande, vous souhaiterez peut-être spécifier quel backend utiliser juste après avoir importé matplotlib et avant de tracer quoi que ce soit. Par exemple, pour utiliser le backend Tk, exécutez `matplotlib.use("TKAgg")`.
> Cependant, dans un notebook Jupyter, les choses sont plus simples : l'importation de `import matplotlib.pyplot` enregistre automatiquement Jupyter lui-même en tant que backend, de sorte que les graphiques s'affichent directement dans le notebook. Auparavant, il nécessitait l'exécution de `%matplotlib inline`. Vous le verrez donc toujours dans certains ordinateurs portables, mais ce n'est plus nécessaire.

Si la fonction `plot` reçoit un tableau de données, elle l'utilisera comme coordonnées sur l'axe vertical et utilisera simplement l'index de chaque point de données dans le tableau comme coordonnée horizontale.
Vous pouvez également fournir deux tableaux : un pour l'axe horizontal `x` et le second pour l'axe vertical `y` :

```python
plt.plot([-3, -2, 5, 0], [1, 6, 4, 3])
plt.show()
```

Les axes correspondent automatiquement à l'étendue des données. Nous aimerions donner un peu plus de place au graphique, appelons donc la fonction `axis` pour modifier l'étendue de chaque axe `[xmin, xmax, ymin, ymax]`.

```python
plt.plot([-3, -2, 5, 0], [1, 6, 4, 3])
plt.axis([-4, 6, 0, 7])
plt.show()
```

Maintenant, traçons une fonction mathématique. Nous utilisons la fonction `linspace` de NumPy pour créer un tableau `x` contenant 500 flottants allant de -2 à 2, puis nous créons un deuxième tableau `y` calculé comme le carré de `x` (pour en savoir plus sur NumPy, lisez le [NumPy tutorial](tools_numpy.ipynb)).

```python
import numpy as np

x = np.linspace(-2, 2, 500)
y = x**2

plt.plot(x, y)
plt.show()
```

C'est un peu sec, ajoutons un titre, des étiquettes x et y, et dessinons une grille.

```python
plt.plot(x, y)
plt.title("Square function")
plt.xlabel("x")
plt.ylabel("y = x**2")
plt.grid(True)
plt.show()
```

# Style et couleur de ligne

Par défaut, matplotlib trace une ligne entre des points consécutifs.

```python
plt.plot([0, 100, 100, 0, 0, 100, 50, 0, 100],
         [0, 0, 100, 100, 0, 100, 130, 100, 0])
plt.axis([-10, 110, -10, 140])
plt.show()
```

Vous pouvez passer le 3ème argument pour changer le style et la couleur de la ligne.
Par exemple, `"g--"` signifie « ligne pointillée verte ».

```python
plt.plot([0, 100, 100, 0, 0, 100, 50, 0, 100],
         [0, 0, 100, 100, 0, 100, 130, 100, 0],
         "g--")
plt.axis([-10, 110, -10, 140])
plt.show()
```

Vous pouvez tracer plusieurs lignes sur un graphique très simplement : il suffit de transmettre `x1, y1, [style1], x2, y2, [style2], ...`.

Par exemple :

```python
plt.plot([0, 100, 100, 0, 0], [0, 0, 100, 100, 0], "r-",
         [0, 100, 50, 0, 100], [0, 100, 130, 100, 0], "g--")
plt.axis([-10, 110, -10, 140])
plt.show()
```

Ou appelez simplement `plot` plusieurs fois avant d'appeler `show`.

```python
plt.plot([0, 100, 100, 0, 0], [0, 0, 100, 100, 0], "r-")
plt.plot([0, 100, 50, 0, 100], [0, 100, 130, 100, 0], "g--")
plt.axis([-10, 110, -10, 140])
plt.show()
```

Vous pouvez également dessiner des points simples au lieu de lignes. Voici un exemple avec des tirets verts, une ligne pointillée rouge et des triangles bleus.
Consultez [the documentation](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib.pyplot.plot) pour la liste complète des options de style et de couleur.

```python
x = np.linspace(-1.4, 1.4, 30)
plt.plot(x, x, 'g--', x, x**2, 'r:', x, x**3, 'b^')
plt.show()
```

La fonction plot renvoie une liste d'objets `Line2D` (un pour chaque ligne). Vous pouvez définir des propriétés supplémentaires sur ces lignes, telles que la largeur de ligne, le style de tiret ou le niveau alpha. Consultez la liste complète des propriétés à [the documentation](https://matplotlib.org/stable/tutorials/introductory/pyplot.html#controlling-line-properties).

```python
x = np.linspace(-1.4, 1.4, 30)
line1, line2, line3 = plt.plot(x, x, 'g--', x, x**2, 'r:', x, x**3, 'b^')
line1.set_linewidth(3.0)
line1.set_dash_capstyle("round")
line3.set_alpha(0.2)
plt.show()
```

# Sous-parcelles
Une figure matplotlib peut contenir plusieurs sous-intrigues. Ces sous-parcelles sont organisées en grille. Pour créer un sous-tracé, appelez simplement la fonction `subplot` et spécifiez le nombre de lignes et de colonnes dans la figure, ainsi que l'index du sous-tracé sur lequel vous souhaitez dessiner (en commençant par 1, puis de gauche à droite et de haut en bas). Notez que pyplot garde une trace du sous-tracé actuellement actif (auquel vous pouvez obtenir une référence en appelant `plt.gca()`), donc lorsque vous appelez la fonction `plot`, elle s'appuie sur le sous-tracé *actif*.

```python
x = np.linspace(-1.4, 1.4, 30)
plt.subplot(2, 2, 1)  # 2 rows, 2 columns, 1st subplot = top left
plt.plot(x, x)
plt.subplot(2, 2, 2)  # 2 rows, 2 columns, 2nd subplot = top right
plt.plot(x, x**2)
plt.subplot(2, 2, 3)  # 2 rows, 2 columns, 3rd subplot = bottom left
plt.plot(x, x**3)
plt.subplot(2, 2, 4)  # 2 rows, 2 columns, 4th subplot = bottom right
plt.plot(x, x**4)
plt.show()
```

* Notez que `subplot(223)` est un raccourci pour `subplot(2, 2, 3)`.

Il est facile de créer des sous-tracés qui s'étendent sur plusieurs cellules de la grille, comme ceci :

```python
plt.subplot(2, 2, 1)  # 2 rows, 2 columns, 1st subplot = top left
plt.plot(x, x)
plt.subplot(2, 2, 2)  # 2 rows, 2 columns, 2nd subplot = top right
plt.plot(x, x**2)
plt.subplot(2, 1, 2)  # 2 rows, *1* column, 2nd subplot = bottom
plt.plot(x, x**3)
plt.show()
```

Si vous avez besoin d'un positionnement de sous-tracé plus complexe, vous pouvez utiliser `subplot2grid` au lieu de `subplot`. Vous spécifiez le nombre de lignes et de colonnes dans la grille, puis la position de votre sous-tracé dans cette grille (en haut à gauche = (0,0)) et éventuellement le nombre de lignes et/ou de colonnes qu'il s'étend. Par exemple:

```python
plt.subplot2grid((3,3), (0, 0), rowspan=2, colspan=2)
plt.plot(x, x**2)
plt.subplot2grid((3,3), (0, 2))
plt.plot(x, x**3)
plt.subplot2grid((3,3), (1, 2), rowspan=2)
plt.plot(x, x**4)
plt.subplot2grid((3,3), (2, 0), colspan=2)
plt.plot(x, x**5)
plt.show()
```

Si vous avez besoin d'encore plus de flexibilité dans le positionnement des sous-intrigues, consultez le [corresponding matplotlib tutorial](https://matplotlib.org/stable/tutorials/intermediate/arranging_axes.html).

# Plusieurs chiffres
Il est également possible de dessiner plusieurs figures. Chaque figure peut contenir une ou plusieurs sous-intrigues. Par défaut, matplotlib crée automatiquement `figure(1)`. Lorsque vous changez de figure, pyplot garde la trace de la figure actuellement active (à laquelle vous pouvez obtenir une référence en appelant `plt.gcf()`), et la sous-intrigue active de cette figure devient la sous-intrigue actuelle.

```python
x = np.linspace(-1.4, 1.4, 30)

plt.figure(1)
plt.subplot(211)
plt.plot(x, x**2)
plt.title("Square and Cube")
plt.subplot(212)
plt.plot(x, x**3)

plt.figure(2, figsize=(10, 5))
plt.subplot(121)
plt.plot(x, x**4)
plt.title("y = x**4")
plt.subplot(122)
plt.plot(x, x**5)
plt.title("y = x**5")

plt.figure(1)      # back to figure 1, current subplot is 212 (bottom)
plt.plot(x, -x**3, "r:")

plt.show()
```

# Machine à états de Pyplot : implicite *vs* explicite
Jusqu'à présent, nous avons utilisé la machine à états de Pyplot qui assure le suivi de la sous-intrigue actuellement active. Chaque fois que vous appelez la fonction `plot`, pyplot s'appuie simplement sur le sous-tracé actuellement actif. Il fait également un peu plus de magie, comme créer automatiquement une figure et une sous-intrigue lorsque vous appelez `plot`, si elles n'existent pas encore. Cette magie est pratique dans un environnement interactif (comme Jupyter).

Mais lorsque vous écrivez un programme, *explicite vaut mieux qu'implicite*. Le code explicite est généralement plus facile à déboguer et à maintenir, et si vous ne me croyez pas, lisez simplement la deuxième règle dans le Zen de Python :

```python
import this
```

Heureusement, Pyplot vous permet d'ignorer complètement la machine à états, ce qui vous permet d'écrire un code magnifiquement explicite. Appelez simplement la fonction `subplots` et utilisez l'objet figure et la liste des objets axes renvoyés. Plus de magie ! Par exemple:

```python
x = np.linspace(-2, 2, 200)
fig1, (ax_top, ax_bottom) = plt.subplots(2, 1, sharex=True)
fig1.set_size_inches(10,5)
line1, line2 = ax_top.plot(x, np.sin(3*x**2), "r-", x, np.cos(5*x**2), "b-")
line3, = ax_bottom.plot(x, np.sin(3*x), "r-")
ax_top.grid(True)

fig2, ax = plt.subplots(1, 1)
ax.plot(x, x**2)
plt.show()
```

Par souci de cohérence, nous continuerons à utiliser la machine à états de pyplot dans la suite de ce tutoriel, mais nous vous recommandons d'utiliser l'interface orientée objet dans vos programmes.

# Pylab *vs* Pyplot *vs* Matplotlib

Il existe une certaine confusion autour de la relation entre pylab, pyplot et matplotlib. C'est simple : matplotlib est la bibliothèque complète, elle contient tout y compris pylab et pyplot.

Pyplot fournit un certain nombre d'outils pour tracer des graphiques, y compris l'interface machine à états avec la bibliothèque de traçage orienté objet sous-jacente.

Pylab est un module pratique qui importe matplotlib.pyplot et NumPy dans un seul espace de noms. Vous trouverez de nombreux exemples utilisant pylab, mais il s'agit désormais de [strongly discouraged](https://matplotlib.org/stable/api/index.html#module-pylab) (car les importations *explicites* sont meilleures que les importations *implicites*).

# Dessiner du texte
Vous pouvez appeler `text` pour ajouter du texte à n’importe quel emplacement du graphique. Spécifiez simplement les coordonnées horizontales et verticales et le texte, et éventuellement quelques arguments supplémentaires. Tout texte dans matplotlib peut contenir des expressions d'équation TeX, voir [the documentation](https://matplotlib.org/stable/tutorials/text/mathtext.html) pour plus de détails.

```python
x = np.linspace(-1.5, 1.5, 30)
px = 0.8
py = px**2

plt.plot(x, x**2, "b-", px, py, "ro")

plt.text(0, 1.5, "Square function\n$y = x^2$", fontsize=20, color='blue',
         horizontalalignment="center")
plt.text(px - 0.08, py, "Beautiful point", ha="right", weight="heavy")
plt.text(px + 0.05, py - 0.4, "x = %0.2f\ny = %0.2f"%(px, py), rotation=-30,
         color='gray')

plt.show()
```

* Remarque : `ha` est un alias pour `horizontalalignment`.

Pour plus de propriétés de texte, visitez [the documentation](https://matplotlib.org/stable/tutorials/text/text_props.html).

De temps en temps, il est nécessaire d’annoter des éléments d’un graphique, comme le beau point ci-dessus. La fonction `annotate` facilite les choses : il suffit d'indiquer l'emplacement du point d'intérêt et la position du texte, plus éventuellement quelques arguments supplémentaires pour le texte et la flèche.

```python
plt.plot(x, x**2, px, py, "ro")
plt.annotate("Beautiful point", xy=(px, py), xytext=(px-1.3,py+0.5),
                           color="green", weight="heavy", fontsize=14,
                           arrowprops={"facecolor": "lightgreen"})
plt.show()
```

Vous pouvez également ajouter un cadre de délimitation autour de votre texte en utilisant l'argument `bbox` :

```python
plt.plot(x, x**2, px, py, "ro")

bbox_props = dict(boxstyle="rarrow,pad=0.3", ec="b", lw=2, fc="lightblue")
plt.text(px-0.2, py, "Beautiful point", bbox=bbox_props, ha="right")

bbox_props = dict(boxstyle="round4,pad=1,rounding_size=0.2", ec="black",
                  fc="#EEEEFF", lw=5)
plt.text(0, 1.5, "Square function\n$y = x^2$", fontsize=20, color='black',
         ha="center", bbox=bbox_props)

plt.show()
```

Juste pour le plaisir, si vous voulez un tracé de style [xkcd](https://xkcd.com), dessinez simplement dans une section `with plt.xkcd()` :

```python
with plt.xkcd():
    plt.plot(x, x**2, px, py, "ro")

    bbox_props = dict(boxstyle="rarrow,pad=0.3", ec="b", lw=2, fc="lightblue")
    plt.text(px-0.2, py, "Beautiful point", bbox=bbox_props, ha="right")

    bbox_props = dict(boxstyle="round4,pad=1,rounding_size=0.2", ec="black",
                      fc="#EEEEFF", lw=5)
    plt.text(0, 1.5, "Square function\n$y = x^2$", fontsize=20, color='black',
             ha="center", bbox=bbox_props)

    plt.show()
```

# Légendes

# Échelles non linéaires
Matplotlib prend en charge les échelles non linéaires, telles que les échelles logarithmiques ou logit.

```python
x = np.linspace(0.1, 15, 500)
y = x**3/np.exp(2*x)

plt.figure(1)
plt.plot(x, y)
plt.yscale('linear')
plt.title('linear')
plt.grid(True)

plt.figure(2)
plt.plot(x, y)
plt.yscale('log')
plt.title('log')
plt.grid(True)

plt.figure(3)
plt.plot(x, y)
plt.yscale('logit')
plt.title('logit')
plt.grid(True)

plt.figure(4)
plt.plot(x, y - y.mean())
plt.yscale('symlog', linthresh=0.05)
plt.title('symlog')
plt.grid(True)

plt.show()
```

# Tiques et tickers
Les axes portent de petites marques appelées « ticks ». Pour être précis, les « ticks » sont les *emplacements* des marques (par exemple (-1, 0, 1)), les « lignes de ticks » sont les petites lignes tracées à ces emplacements, les « étiquettes de ticks » sont les étiquettes dessinées à côté des lignes de ticks et les « tickers » sont des objets capables de décider où placer les ticks. Les tickers par défaut font généralement un très bon travail en plaçant environ 5 à 8 ticks à une distance raisonnable les uns des autres.

Mais parfois, vous avez besoin de plus de contrôle (par exemple, il y a trop d'étiquettes de graduations sur le graphique logit ci-dessus). Heureusement, matplotlib vous donne un contrôle total sur les ticks. Vous pouvez même activer des ticks mineurs.

```python
x = np.linspace(-2, 2, 100)

plt.figure(1, figsize=(15,10))
plt.subplot(131)
plt.plot(x, x**3)
plt.grid(True)
plt.title("Default ticks")

ax = plt.subplot(132)
plt.plot(x, x**3)
ax.xaxis.set_ticks(np.arange(-2, 2, 1))
plt.grid(True)
plt.title("Manual ticks on the x-axis")

ax = plt.subplot(133)
plt.plot(x, x**3)
plt.minorticks_on()
ax.tick_params(axis='x', which='minor', bottom=False)
ax.xaxis.set_ticks([-2, 0, 1, 2])
ax.yaxis.set_ticks(np.arange(-5, 5, 1))
ax.yaxis.set_ticklabels(["min", -4, -3, -2, -1, 0, 1, 2, 3, "max"])
plt.grid(True)
plt.title("Manual ticks and tick labels\n(plus minor ticks) on the y-axis")

plt.show()
```

# Projection polaire
Dessiner un graphique polaire est aussi simple que de définir l'argument `projection` sur `"polar"` lors de la création du sous-tracé.

```python
radius = 1
theta = np.linspace(0, 2*np.pi*radius, 1000)

plt.subplot(111, projection='polar')
plt.plot(theta, np.sin(5*theta), "g-")
plt.plot(theta, 0.5*np.cos(20*theta), "b-")
plt.show()
```

# Projection 3D

Le traçage de graphiques 3D est assez simple : lors de la création d'un sous-tracé, définissez `projection` sur `"3d"`. Il renvoie un objet axes 3D, que vous pouvez utiliser pour appeler `plot_surface`, fournissant les coordonnées x, y et z, ainsi que d'autres arguments facultatifs. Pour plus d'informations sur la génération de tracés 3D, consultez le [matplotlib tutorial](https://matplotlib.org/stable/tutorials/toolkits/mplot3d.html).

```python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

figure = plt.figure(1, figsize = (12, 4))
subplot3d = plt.subplot(111, projection='3d')
surface = subplot3d.plot_surface(X, Y, Z, rstride=1, cstride=1,
                                 cmap=matplotlib.cm.coolwarm, linewidth=0.1)
plt.show()
```

Une autre façon d'afficher ces mêmes données est *via* un tracé de contour.

```python
plt.contourf(X, Y, Z, cmap=matplotlib.cm.coolwarm)
plt.colorbar()
plt.show()
```

# Lignes
Vous pouvez tracer des lignes simplement en utilisant la fonction `plot`, comme nous l'avons fait jusqu'à présent. Cependant, il est souvent pratique de créer une fonction utilitaire qui trace une ligne (apparemment) infinie sur le graphique, étant donné une pente et une intersection. Vous pouvez également utiliser les fonctions `hlines` et `vlines` qui tracent des segments de ligne horizontaux et verticaux.
Par exemple:

```python
def plot_line(axis, slope, intercept, **kwargs):
    xmin, xmax = axis.get_xlim()
    plt.plot([xmin, xmax],
             [xmin*slope+intercept, xmax*slope+intercept],
             **kwargs)

x = np.random.randn(1000)
y = 0.5*x + 5 + np.random.randn(1000) * 2
plt.axis([-2.5, 2.5, -5, 15])
plt.scatter(x, y, alpha=0.2)
plt.plot(1, 0, "ro")
plt.vlines(1, -5, 0, color="red")
plt.hlines(0, -2.5, 1, color="red")
plot_line(axis=plt.gca(), slope=0.5, intercept=5, color="magenta")
plt.grid(True)
plt.show()
```

# Images
Lire, générer et tracer des images dans matplotlib est assez simple.

Pour lire une image, il suffit d'importer le module `matplotlib.image`, et d'appeler sa fonction `imread`, en lui passant le nom du fichier (ou l'objet fichier). Il renvoie les données de l'image, sous forme de tableau NumPy. Essayons avec l'image `my_square_function.png` que nous avons enregistrée plus tôt.

```python
import matplotlib.image as mpimg

img = mpimg.imread("my_square_function.png")
print(img.shape, img.dtype)
```

Nous avons chargé une image 288x432. Chaque pixel est représenté par un tableau de 4 éléments : niveaux rouge, vert, bleu et alpha, stockés sous forme de flottants 32 bits entre 0 et 1. Il ne nous reste plus qu'à appeler `imshow` :

```python
plt.imshow(img)
plt.show()
```

Tadaaa ! Vous souhaiterez peut-être masquer les axes lorsque vous affichez une image :

```python
plt.imshow(img)
plt.axis('off')
plt.show()
```

Sous le capot, `imread()` utilise la bibliothèque d'images Python (PIL), et la documentation de Matplotlib recommande désormais d'utiliser directement PIL :

```python
import PIL

img = np.asarray(PIL.Image.open("my_square_function.png"))
print(img.shape, img.dtype)
```

Notez que le tableau contient désormais des entiers non signés de 8 bits (de 0 à 255), mais ce n'est pas grave car `plt.imshow()` prend également en charge ce format :

```python
plt.imshow(img)
plt.axis('off')
plt.show()
```

Vous pouvez également générer une image à partir de zéro :

```python
img = np.arange(100*100).reshape(100, 100)
print(img)
plt.imshow(img)
plt.show()
```

Comme nous n'avons pas fourni de niveaux RVB, la fonction `imshow` mappe automatiquement les valeurs sur un dégradé de couleurs. Par défaut, le dégradé de couleurs va du bleu (pour les valeurs faibles) au jaune (pour les valeurs élevées), mais vous pouvez sélectionner une autre palette de couleurs. Par exemple:

```python
plt.imshow(img, cmap="hot")
plt.show()
```

Vous pouvez également générer directement une image RVB :

```python
img = np.empty((20,30,3))
img[:, :10] = [0, 0, 0.6]
img[:, 10:20] = [1, 1, 1]
img[:, 20:] = [0.6, 0, 0]
plt.imshow(img)
plt.show()
```

Étant donné que le tableau `img` est assez petit (20 x 30), lorsque la fonction `imshow` l'affiche, l'image s'agrandit à la taille de la figure. Imaginez étirer l'image originale, en laissant des espaces entre les pixels d'origine. Comment imshow remplit-il les vides ? Eh bien, par défaut, il colore simplement chaque pixel vide en utilisant la couleur du pixel non vide le plus proche. Cette technique peut conduire à des images pixellisées. Si vous préférez, vous pouvez utiliser une méthode d'interpolation différente, telle que [bilinear interpolation](https://en.wikipedia.org/wiki/Bilinear_interpolation) pour remplir les pixels vides. Cela conduit à des bords flous, ce qui peut être plus agréable dans certains cas :

```python
plt.imshow(img, interpolation="bilinear")
plt.show()
```

#Animations
Bien que matplotlib soit principalement utilisé pour générer des images, il est également capable d'afficher des animations. Tout d’abord, vous devez importer `matplotlib.animation`.

```python
import matplotlib.animation as animation
```

Dans l'exemple suivant, nous commençons par créer des points de données, puis nous créons un tracé vide, nous définissons la fonction de mise à jour qui sera appelée à chaque itération de l'animation, et enfin nous ajoutons une animation au tracé en créant une instance `FuncAnimation`.

Le constructeur `FuncAnimation` prend une figure, une fonction de mise à jour et des arguments facultatifs. Nous précisons que nous souhaitons une animation de 50 images, avec 100 ms entre chaque image. A chaque itération, `FuncAnimation` appelle notre fonction de mise à jour et lui passe le numéro de trame `num` (de 0 à 49 dans notre cas) suivi des arguments supplémentaires que nous avons spécifiés avec `fargs`.

Notre fonction de mise à jour définit simplement les données de ligne comme étant les premiers points de données `num` (de sorte que les données soient dessinées progressivement), et juste pour le plaisir, nous ajoutons également un petit nombre aléatoire à chaque point de données afin que la ligne semble bouger.

```python
x = np.linspace(-1, 1, 100)
y = np.sin(x**2*25)
data = np.array([x, y])

fig = plt.figure()
line, = plt.plot([], [], "r-")  # start with an empty plot
plt.axis([-1.1, 1.1, -1.1, 1.1])
plt.plot([-0.5, 0.5], [0, 0], "b-", [0, 0], [-0.5, 0.5], "b-", 0, 0, "ro")
plt.grid(True)
plt.title("Marvelous animation")

# this function will be called at every iteration
def update_line(num, data, line):
    # we only plot the first `num` data points.
    line.set_data(data[..., :num] + np.random.rand(2, num) / 25)
    return line,

line_ani = animation.FuncAnimation(fig, update_line, frames=50,
                                   fargs=(data, line), interval=100)
plt.close()  # call close() to avoid displaying the static plot
```

Ensuite, affichons l'animation. Une option consiste à le convertir en code HTML5 (à l'aide d'une balise `<video>`) et à restituer ce code à l'aide de `IPython.display.HTML` :

```python
from IPython.display import HTML

HTML(line_ani.to_html5_video())
```

Alternativement, nous pouvons afficher l'animation à l'aide d'un joli petit widget interactif HTML/Javascript :

```python
HTML(line_ani.to_jshtml())
```

Vous pouvez configurer Matplotlib pour utiliser ce widget par défaut lors du rendu des animations :

```python
matplotlib.rc('animation', html='jshtml')
```

Après cela, vous n'avez même plus besoin d'utiliser `IPython.display.HTML` :

```python
animation.FuncAnimation(fig, update_line, frames=50, fargs=(data, line),
                        interval=100)
```

**Attention :** si vous enregistrez le bloc-notes avec ses sorties, les animations prendront beaucoup de place.

# Enregistrer des animations dans des fichiers vidéo
Matplotlib s'appuie sur des bibliothèques tierces pour écrire des vidéos telles que [FFMPEG](https://www.ffmpeg.org/) ou [ImageMagick](https://imagemagick.org/). Dans l'exemple suivant, nous utiliserons FFMPEG, alors assurez-vous de l'installer en premier. Pour enregistrer l'animation au format GIF, vous aurez besoin d'ImageMagick.

```python
Writer = animation.writers['ffmpeg']
writer = Writer(fps=15, metadata=dict(artist='Me'), bitrate=1800)
line_ani.save('my_wiggly_animation.mp4', writer=writer)
```
