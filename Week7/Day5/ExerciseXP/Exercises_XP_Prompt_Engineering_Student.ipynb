## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
# Exercise 1: Identifying issues with "Write something about productivity tips."
critical_issues = [
    'Lack of Context: It does not specify the audience or the product (FlowNest).',
    'Ambiguous Format: It doesn\'t define if it should be a list, an essay, or a short post.',
    'No Tone or Voice: It doesn\'t specify if the brand is professional, quirky, or urgent.'
]
critical_issues
```

```python
# Exercise 1: Rewriting with Role, Instructions, and Constraints
rewritten_prompt = """
Role: You are an expert Social Media Manager for FlowNest, a productivity app for remote teams.
Task: Write a high-engagement LinkedIn post providing 3 actionable productivity tips for managers.
Constraints:
- Tone: Professional yet encouraging.
- Format: Use bullet points for tips.
- Length: Maximum 150 words.
- Include: A call-to-action to check out FlowNest.
- Reusability: To reuse this for future posts, simply swap the 'Task' topic and the 'Tips' count.
"""
rewritten_prompt
```

## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# Exercise 2: Multi-Part Prompt for Quiz Generation
quiz_prompt = """
Context: You are a middle school science teacher.
Topic: Volcanic Eruptions.
Target Audience: 7th-grade students (11-13 years old).
Instructions:
1. Provide a 2-bullet point summary of how volcanoes erupt using simple vocabulary.
2. Create 3 Multiple Choice Questions (MCQs). Each must have 1 correct answer and 2 plausible distractors.
Output Format: Format the output as clear headers and lists so it can be easily copied into Google Slides.
Tone: Friendly and educational.
"""
quiz_prompt
```

```python
# Exercise 2: Explaining improvements
improvements = [
    'Audience Calibration: Uses age-appropriate vocabulary (11-13 y/o) instead of generic "kids".',
    'Structural Control: Explicitly requests a 2-bullet summary and 3 MCQs to fit slide layouts.',
    'Negative Constraint Management: Defines distractors to ensure the quiz is actually challenging but fair.'
]
improvements
```
