### My answer

#### Problem Statement
The objective of this project is to predict whether a loan applicant will **default** on a loan (a binary target variable, `default` = 1 if the borrower fails to repay according to the agreed terms, 0 otherwise), so that the lending institution can support the decision of whether to **approve, reject, or price a loan application** and manage its overall credit risk exposure.

#### Data Types to Collect
- **Applicant personal and demographic details** (age, employment length, employment status, marital status): these provide context on the borrower's stability and capacity to generate steady income, which influences their ability to repay.
- **Financial information** (annual income, debt-to-income ratio, existing liabilities): this directly measures how much disposable income the applicant has available to service new debt, which is one of the strongest indicators of repayment ability.
- **Credit history and scores** (credit score, number of past delinquencies, credit utilization, length of credit history): past repayment behaviour is one of the best predictors of future repayment behaviour, so this strongly correlates with default risk.
- **Loan characteristics** (loan amount, interest rate, term, loan purpose): larger amounts, higher rates, and longer terms generally increase the repayment burden and therefore the probability of default.
- **Behavioural and repayment history** (past loan repayments, missed payments, account activity): a documented pattern of late or missed payments signals elevated risk for the new loan.

#### Data Sources
- **The financial institution's internal records**, which contain historical loan applications, account balances, and repayment outcomes, and which can be obtained directly from the bank's core banking and loan-management databases.
- **Credit bureaus** (such as Experian, Equifax, or TransUnion), which provide standardized credit scores and cross-institution credit history, and which are typically integrated through a licensed API or a periodic data feed.
- **Public and regulatory data sources** (such as census income statistics or regional economic indicators), which can be merged on geographic keys like state or zip code to add macro-economic context.
- **Application-form data** captured at the point of application, which is collected directly from the customer through the digital or in-branch onboarding process.

#### Risks and Constraints
Collecting and using this data raises several concerns that must be managed carefully: privacy and regulation require compliance with laws such as GDPR and fair-lending rules (e.g., the Equal Credit Opportunity Act), meaning sensitive attributes must be protected and the model must not discriminate against protected groups; data quality issues such as missing values, inconsistent formats, and outdated credit records can degrade model reliability; sampling bias is a real threat because historical data only reflects applicants who were previously approved, which can skew the model toward an unrepresentative population; and strong data governance, including consent management, audit trails, access control, and model explainability, is essential to ensure the system remains ethical, auditable, and trusted by both regulators and customers.

### My justification

The features I selected as most relevant are `credit_score`, `debt_to_income`, `annual_income`, `loan_amount`, `interest_rate`, `employment_length`, `num_delinquencies`, `total_utilization`, `num_open_accounts`, `home_ownership`, `purpose`, and `term`.

**Justification of each selected feature:**
- **`credit_score`** summarizes a borrower's overall credit reliability, so a lower score is strongly associated with a higher probability of default.
- **`debt_to_income`** measures how much of the applicant's income is already committed to debt, and a higher ratio means less room to absorb a new payment, raising default risk.
- **`annual_income`** reflects raw repayment capacity, because higher and more reliable income makes it easier to keep up with scheduled payments.
- **`loan_amount`** captures the size of the obligation, and larger loans demand larger payments that are harder to sustain, increasing default likelihood.
- **`interest_rate`** both raises the monthly burden and acts as the lender's own pricing of risk, so higher rates tend to flag riskier borrowers.
- **`employment_length`** is a proxy for income stability, since longer tenure suggests a steadier cash flow available for repayment.
- **`num_delinquencies`** directly records past missed or late payments, and past delinquency is one of the strongest predictors of future default.
- **`total_utilization`** shows how stretched the borrower's existing credit already is, and high utilization signals financial stress.
- **`num_open_accounts`** indicates the breadth of current obligations, which together with utilization helps gauge how leveraged the applicant is.
- **`home_ownership`** provides a signal of financial stability and potential asset backing, which can lower default risk.
- **`purpose`** matters because default rates differ systematically by loan purpose (for example, debt consolidation versus a small-business loan).
- **`term`** captures loan duration, and longer terms expose the borrower to more time and uncertainty, which tends to increase cumulative default risk.

**Excluded features:**
- **`state`** is excluded because, on its own, it is a coarse geographic label that contributes little direct predictive signal and risks introducing regional bias.
- **`zip_code`** is excluded because it is a very high-cardinality identifier that can leak location-based proxies for protected attributes and easily lead to overfitting.
- **`age`** is excluded as a direct model feature because using it can raise fair-lending and discrimination concerns under regulations such as the Equal Credit Opportunity Act.
- **`application_type`** is excluded only if it shows little variation in the data, since a near-constant feature adds noise without improving prediction.

**Encoding and imputation:** I would encode categorical features such as `home_ownership`, `purpose`, and `term` using one-hot encoding for low-cardinality variables (and ordinal or target encoding for higher-cardinality ones), fitting the encoder only on the training set to avoid leakage. I would impute missing values using the median for numeric features (which is robust to outliers like extreme incomes) and the most frequent category for categorical features, again fitting the imputer on the training data and optionally adding a binary "was-missing" indicator column.

### Instructions
Which model(s) would you pick for a Loan Prediction ?

### Guidance
- Begin by naming one or two candidate models in a complete sentence and explain why each model is suitable for this problem.

```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

### My answer

**Model choice.** For loan default prediction I would start with **Logistic Regression** as a baseline because it is fast, well-calibrated, and highly interpretable, which matters in a regulated lending context where each decision must be explainable, and I would then move to a **Gradient Boosting model (such as XGBoost or LightGBM)** because it captures non-linear interactions between features and typically delivers stronger predictive performance on tabular financial data.

**Evaluation plan.** I would first split the data into a training set and a held-out test set using **stratified sampling** so that the proportion of defaults is preserved in both sets, and within the training set I would use **stratified k-fold cross-validation** (for example, 5 folds) to obtain a stable estimate of generalization performance. Because default prediction is imbalanced and the cost of a missed default is high, I would not rely on accuracy alone; instead I would report **precision, recall, F1-score, ROC-AUC, and especially PR-AUC (average precision)**, since the precision–recall curve is more informative when the positive class is rare. I would choose the **decision threshold** not at the default 0.5 but at the point on the validation precision–recall curve that best matches the business cost trade-off, accepting more false positives if missing a true default is far more expensive.

**Class imbalance.** I would address the imbalance between non-defaulters and defaulters by using **stratification** in every split, applying **class weights** (for example, `class_weight="balanced"` or `scale_pos_weight` in boosting) so the model penalizes errors on the minority class more heavily, and, if needed, experimenting with **resampling techniques such as SMOTE or random undersampling** applied only inside the cross-validation training folds to avoid leaking information.

**Hyperparameter iteration.** I would tune hyperparameters using a grid or randomized/Bayesian search **wrapped inside the cross-validation loop** (via a scikit-learn `Pipeline`), so that all preprocessing, encoding, imputation, and resampling are fit only on the training portion of each fold, which prevents data leakage and ensures the reported performance is an honest estimate of how the model will behave on unseen applicants.

### My answer

**Predicting Stock Prices — Supervised Learning (Regression).** This scenario is best solved with supervised learning, specifically regression (and often time-series models such as ARIMA or LSTM networks), because we want to predict a continuous numeric value from historically labelled data. The input data consists of historical price series together with technical indicators, trading volume, and relevant economic or news signals. The output is the predicted future price (or return) as a continuous number. The learning signal is the difference between the model's predicted price and the actual observed price, which the model minimizes through a regression loss such as mean squared error.

**Organizing a Library of Books — Unsupervised Learning (Clustering).** This scenario is best solved with unsupervised learning, specifically clustering, because there are no predefined genre labels and the goal is to discover natural groupings based on similarity. The input data consists of features extracted from each book, such as text content, keywords, author, and topic embeddings. The output is a set of clusters that group similar books together into emergent categories. The learning signal is internal rather than supervised: the algorithm optimizes a measure of within-cluster similarity and between-cluster separation, since no ground-truth labels are provided.

**Robot Navigating a Maze — Reinforcement Learning.** This scenario is best solved with reinforcement learning, because an agent must learn a sequence of actions through trial-and-error interaction with its environment rather than from a labelled dataset. The input data is the robot's current state, such as its position in the maze and its sensor readings. The output is the action the robot takes at each step (for example, move up, down, left, or right). The learning signal is the reward received from the environment, where reaching the exit or taking shorter paths yields positive reward and hitting walls or wasting steps yields penalties, so the agent learns a policy that maximizes cumulative reward and thereby finds the shortest path.

**Unsupervised Learning (a clustering model, e.g., K-Means).** Because there are no ground-truth labels, I would assess cluster quality with **internal validation metrics** such as the **silhouette score** (higher means clusters are dense and well separated), the **Davies–Bouldin index** (lower is better), and the **Calinski–Harabasz score**, and I would use the **elbow method** by plotting within-cluster sum of squares (inertia) against the number of clusters to identify the point where adding clusters yields diminishing returns.
