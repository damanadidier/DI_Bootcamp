# Machine Learning: A Summary Document

## 1. The Basics of Machine Learning — and Why It Matters for Data Analysts

**Machine Learning (ML)** is a branch of Artificial Intelligence that gives computers the ability to *learn patterns from data* and make predictions or decisions **without being explicitly programmed** for every rule. Instead of a developer hand-coding instructions ("if X then Y"), an ML algorithm is shown many examples and infers the rules on its own.

At its core, ML works by:
1. Taking in **data** (the examples).
2. Identifying **patterns / relationships** within that data through a training process.
3. Producing a **model** that can generalize those patterns to *new, unseen* data.

**Why it is important for data analysts:**
- **Scale & speed** — ML can find patterns across millions of records far faster than manual analysis.
- **Prediction, not just description** — traditional analytics describes *what happened*; ML helps forecast *what will happen* (e.g., next quarter's sales, churn risk).
- **Automation of repetitive decisions** — classification, scoring, and segmentation can be automated, freeing analysts for higher-value work.
- **Deeper insight** — ML surfaces non-obvious relationships in data that simple statistics might miss.
- **Better decision support** — it turns raw data into actionable, evidence-based recommendations for the business.

## 3. The Three Main Types of Machine Learning

### a) Supervised Learning
- **Definition:** The model learns from **labeled data** — each training example has both input features *and* the correct output (the "answer"). The goal is to learn a mapping from inputs to outputs so it can predict labels for new data.
- **Common tasks:** Classification (predicting categories) and Regression (predicting continuous numbers).
- **Example scenario:** **Email spam detection** — the model is trained on thousands of emails already labeled "spam" or "not spam," then predicts the label for incoming emails. Another example: predicting **house prices** from features like size, location, and number of rooms.

### b) Unsupervised Learning
- **Definition:** The model learns from **unlabeled data** — there are no predefined answers. The algorithm tries to discover hidden structure, groupings, or patterns on its own.
- **Common tasks:** Clustering (grouping similar items) and Dimensionality Reduction.
- **Example scenario:** **Customer segmentation** — a retailer groups customers into segments (e.g., budget shoppers, premium buyers, occasional visitors) based on purchasing behavior, without any pre-existing labels, to target marketing campaigns.

### c) Reinforcement Learning
- **Definition:** An **agent** learns by *interacting with an environment* through trial and error. It takes actions, receives **rewards or penalties**, and gradually learns the strategy (policy) that maximizes cumulative reward.
- **Common tasks:** Sequential decision-making and control.
- **Example scenario:** **Game-playing AI / robotics** — a program learning to play chess or Go (e.g., AlphaGo), or a robot learning to walk, improves over time by being rewarded for good moves and penalized for bad ones. Also used in self-driving car navigation.

## 4. The Process of Developing a Machine Learning Model

Developing an ML model is an iterative process. Three of the most important stages are:

### Stage 1 — Feature Selection
- **What it is:** Choosing the most relevant input variables (features) from the dataset that will help the model make accurate predictions.
- **Why it matters:** Irrelevant or redundant features add noise, increase complexity, slow down training, and can cause **overfitting**. Selecting the right features improves accuracy and interpretability.
- **Activities:** Data cleaning, handling missing values, creating new features (feature engineering), and removing low-value or highly correlated features.

### Stage 2 — Model Selection
- **What it is:** Choosing the appropriate **algorithm** for the problem (e.g., Linear Regression, Decision Tree, Random Forest, K-Means, Neural Network).
- **Why it matters:** Different problems and data types suit different algorithms — there is no single "best" model. The choice depends on the task (classification vs. regression vs. clustering), data size, and the trade-off between accuracy and interpretability.
- **Activities:** Splitting data into training/test sets, training candidate models, and tuning hyperparameters.

### Stage 3 — Model Evaluation
- **What it is:** Measuring how well the trained model performs on **unseen data** using appropriate metrics.
- **Why it matters:** Evaluation confirms whether the model generalizes well or is overfitting/underfitting. It guides whether to deploy the model or go back and refine earlier stages.
- **Common metrics:**
  - *Classification:* Accuracy, Precision, Recall, F1-score, Confusion Matrix.
  - *Regression:* Mean Squared Error (MSE), Root Mean Squared Error (RMSE), R² score.

### Summary
Machine Learning enables systems to learn from data and improve over time. For data analysts it unlocks prediction, automation, and deeper insight. Its three core types — **Supervised**, **Unsupervised**, and **Reinforcement Learning** — suit different kinds of problems, and a solid model is built through an iterative cycle of **Feature Selection → Model Selection → Model Evaluation**.
