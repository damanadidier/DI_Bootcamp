## A) Server (server.py)
Implement the WeatherDemo server with one tool and one resource.

````python
%%writefile server.py
import logging
from mcp.server.fastmcp import FastMCP

# Configure logging to see server activity
logging.basicConfig(level=logging.INFO)

# Initialize FastMCP server named 'WeatherDemo'
mcp = FastMCP("WeatherDemo")

# Local data store
CITY_DATA = {
    "paris": {"temp_c": 21, "condition": "sunny"},
    "london": {"temp_c": 18, "condition": "cloudy"},
    "new york": {"temp_c": 24, "condition": "breezy"},
}

@mcp.tool()
def get_weather(city: str) -> dict:
    """Return simple weather info for a supported city."""
    key = city.strip().lower()
    data = CITY_DATA.get(key)
    if not data:
        return {"error": f"Unsupported city: {city}. Try one of: {', '.join(CITY_DATA)}"}

    logging.info("get_weather called for %s", key)
    return {"city": key, **data}

@mcp.resource("cities://list")
def list_cities() -> str:
    """Return supported cities as newline-separated text."""
    # JOIN with newline for better readability in the client
    return "\n".join(sorted(CITY_DATA))

if __name__ == "__main__":
    # Run the server using the built-in transport (defaults to stdio)
    mcp.run()
````

## B) Client (client.py)
Spawn the server via STDIO, discover capabilities, and call them.

````python
%%writefile client.py
import asyncio
import sys
from pathlib import Path
from mcp import ClientSession, StdioServerParameters
from mcp.client.stdio import stdio_client

# Define how to launch the server using the current Python interpreter
SERVER_PATH = Path(__file__).parent / "server.py"
server_params = StdioServerParameters(
    command=sys.executable,
    args=[str(SERVER_PATH)],
    env=None
)

def extract_content(payload):
    """Helper to extract text content from MCP response objects."""
    if hasattr(payload, "contents"):
        contents = payload.contents
        if contents:
            first = contents[0]
            if hasattr(first, "text"): return first.text
            if isinstance(first, dict) and "text" in first: return first["text"]
            return str(first)
    if hasattr(payload, "content"): return payload.content
    return str(payload)

async def run():
    # 1. Establish stdio connection to the server
    async with stdio_client(server_params) as (read, write):
        # 2. Create a session to manage requests/responses
        async with ClientSession(read, write) as session:
            # 3. Initialize the connection
            await session.initialize()

            # 4. Discover what resources the server offers
            resources = await session.list_resources()
            print("--- Resources ---")
            for res in resources.resources:
                print(f"- {res.uri} ({res.name or 'No name'})")

            # 5. Discover what tools the server offers
            tools = await session.list_tools()
            print("\n--- Tools ---")
            for tool in tools.tools:
                print(f"- {tool.name}")

            # 6. Read from a resource (the city list)
            print("\nReading resource 'cities://list':")
            cities = await session.read_resource("cities://list")
            print(extract_content(cities))

            # 7. Call a tool (get weather for Paris)
            print("\nCalling tool 'get_weather' for Paris:")
            weather = await session.call_tool("get_weather", {"city": "Paris"})
            print(extract_content(weather))

if __name__ == "__main__":
    # Start the async client loop
    asyncio.run(run())
````
