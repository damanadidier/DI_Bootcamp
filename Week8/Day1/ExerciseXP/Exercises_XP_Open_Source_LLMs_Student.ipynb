```python
# Exercise 1: Capturing definitions and key characteristics of LLM openness
open_source_levels = {
    "Fully Open": {
        "definition": "Models where the training code, data (often with caveats), weights, and architecture are all released.",
        "what_is_open": ["Weights", "Training Data/Recipes", "Evaluation Code", "Architecture"],
        "what_you_can_do": ["Audit data bias", "Replicate training from scratch", "Complete domain adaptation"],
        "what_you_cannot_do": ["Claim proprietary ownership of the base architecture", "Violate specific license ethics clauses"],
    },
    "Weights Released": {
        "definition": "The pre-trained weights are available for download, but the full training dataset or exact training code might be private.",
        "what_is_open": ["Final model weights", "Inference code"],
        "what_you_can_do": ["Local deployment", "Fine-tuning (LoRA/QLoRA)", "Quantization"],
        "what_you_cannot_do": ["Verify the exact data distribution used for pre-training", "Retrain from step zero"],
    },
    "Architecture Only": {
        "definition": "Only the technical specifications and layers are shared, usually via a paper, without the trained weights.",
        "what_is_open": ["Paper/Technical report", "Network structure/Layers"],
        "what_you_can_do": ["Build a compatible engine", "Understand the theoretical logic"],
        "what_you_cannot_do": ["Use it for inference without massive compute for training", "Run it out-of-the-box"],
    },
}
open_source_levels
```

```python
# Filling the comparison table with technical impact details
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Weights, Code, Data | Enables deep auditing and full retraining if compute allows. |
| Weights Released | Model parameters | Perfect for fine-tuning on custom data but limited data transparency. |
| Architecture Only | Technical specs | High barrier; requires training from scratch (millions of dollars). |"""
print(comparison_table)
```

```python
# 3-5 sentence comparative paragraph on openness
comparative_paragraph = """The transition from 'Fully Open' to 'Architecture Only' models represents a significant shift in user autonomy and transparency. While Weights Released models like Llama are popular for local deployment, Fully Open models like Pythia allow for a deeper understanding of data biases. In contrast, Architecture Only releases serve primarily as academic contributions, offering limited utility to developers without extreme compute budgets for training."""
print(comparative_paragraph)
```
